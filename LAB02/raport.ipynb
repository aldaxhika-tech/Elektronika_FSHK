## 1. Qëllimi i Eksperimentit

Qëllimi i këtij laboratori është analizimi i funksionimit të pjesëtuesit të tensionit me rezistorë të rendit kΩ dhe MΩ, si dhe studimi i tensionit dalës ($V_{out}$) në varësi të vlerave të ndryshme të ngarkesës ($R_3 / R_L$)

## 2. Baza Teorike

* **Pjesëtuesi i tensionit pa ngarkesë:** Për dy rezistorë $R_1$ dhe $R_2$ në seri me tension hyrës $V_{in}$, tensioni te $R_2$ jepet nga:
  $$V_{out} = V_{in} \cdot \frac{R_2}{R_1 + R_2}$$

* **Raporti i tensioneve:** Raporti midis tensionit dalës dhe atij hyrës shprehet si:
  $$\frac{U_{out}}{U_{in}} = \frac{V_{out}}{V_{in}}$$

* **Ndikimi i ngarkesës:** Shtimi i një ngarkese $R_L$ ndryshon rezistencën ekuivalente dhe rrjedhimisht tensionin e matur në dalje.

## 3. Ndërtimi Eksperimental

1. **Eksperimenti 1:** Ndërtimi i pjesëtuesit të tensionit me $R_1 = 2.22\text{ k}\Omega$ dhe $R_2 = 2.18\text{ k}\Omega$ me $V_{in} = 11.28\text{ V}$ dhe matja e $V_{out}$.
2. **Eksperimenti 2:** Ndërtimi i pjesëtuesit me rezistorë të vlerave të larta $R_1 = 3.212\text{ M}\Omega$ dhe $R_2 = 3.250\text{ M}\Omega$ me $V_{in} = 11.28\text{ V}$ dhe matja e $V_{out}$.
3. **Eksperimenti 3 (Matjet 1–5):** Matja e $V_{out}$ për pesë vlera të ndryshme të ngarkesës $R_3 / R_L$: $46\ \Omega$, $100\ \Omega$, $220\ \Omega$, $470\ \Omega$ dhe $1000\ \Omega$.

# 5. Analiza e Te Dhenave
